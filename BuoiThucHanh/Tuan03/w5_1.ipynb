### Introduction
What is Exploratory Data Analysis ?

Exploratory Data Analysis or (EDA) is understanding the data sets by summarizing their main characteristics often plotting them visually. This step is very important especially when we arrive at modeling the data in order to apply Machine learning. Plotting in EDA consists of Histograms, Box plot, Scatter plot and many more. It often takes much time to explore the data. Through the process of EDA, we can ask to define the problem statement or definition on our data set which is very important.

### How to perform Exploratory Data Analysis ?

This is one such question that everyone is keen on knowing the answer. Well, the answer is it depends on the data set that you are working. There is no one method or common methods in order to perform EDA, whereas in this tutorial you can understand some common methods and plots that would be used in the EDA process.


### Types of exploratory data analysis 

There are four types of exploratory data analysis:

1️⃣ Univariate non-graphic. This is the simplest form of data analysis and the data analyzed consists of only one variable. Since it is a single variable, it does not deal with causes or relationships. The main purpose of univariate analysis is to identify data and find patterns that exist in it.

2️⃣ Bivariate, graphic. Non-graphical methods do not provide a complete picture of the data.  Therefore, graphical methods are required.  Common bivariate chart types include: 👇🏻

🔹️ Root and leaf plots showing all data values ​​and the shape of the distribution.

🔹️ Histograms, which are a bar plot where each bar represents the frequency (number) or proportion (number/total number) for a range of values.

🔹️ Box charts that graphically display a five-figure summary of the minimum, first quarter, median, third quarter, and maximum values.

3️⃣ Multivariate non-graphic: Multivariate data arises according to more than one variable.  Multivariate non-graphical EDA techniques show the relationship between two or more variables of the data, often through cross tabulation or statistics.

4️⃣ Multivariate graphic: Multivariate data uses charts to display relationships between two or more data sets. The most commonly used chart is a bar plot or bar chart grouped with each bar within a group representing one level of each group and the levels of the other variable.

Other general types of multivariate graphs are: 

🔸️ Scatterplot used to plot data points on the horizontal and vertical axes to show how much one variable is affected by another variable.

🔸️ Multivariate chart, which is a graphical representation of the relationships between factors and a response.

🔸️ Bubble chart, a data visualization that displays multiple circles (bubbles) in a two-dimensional drawing.

🔸️ Heatmap, which is a graphical representation of data with values ​​represented by colors.

8.Detecting Outliers¶

An outlier is a point or set of points that are different from other points. Sometimes they can be very high or very low. It's often a good idea to detect and remove the outliers. Because outliers are one of the primary reasons for resulting in a less accurate model. Hence it's a good idea to remove them. The outlier detection and removing that I am going to perform is called IQR score technique. Often outliers can be seen with visualizations using a box plot. Shown below are the box plot of MSRP, Cylinders, Horsepower and EngineSize. Herein all the plots, you can find some points are outside the box they are none other than outliers. The technique of finding and removing outlier that I am performing in this assignment is taken help of a tutorial from towards data science.

```python
# TODO: plot a boxplot of Price (Vẽ boxplot của Year (vì ko có Price))
plt.boxplot(df['Year'])
plt.ylabel('Year')
plt.title('Boxplot of Year')
plt.show()
```

```python
# TODO: check HP outliers with a boxplot (Kt các gtri ngoại lai của biến HP bằng boxplot)
plt.boxplot(df['Engine HP'])
plt.ylabel('Engine HP')
plt.title('Boxplot of Engine HP')
plt.show()
```

```python
# TODO: plot a boxplot of Cylinders (Vẽ boxplot của Cylinders)
plt.boxplot(df['Engine Cylinders'])
plt.ylabel('Engine Cylinders')
plt.title('Boxplot of Engine Cylinders')
plt.show()
```

```python
# TODO: compute (tính) Q1, Q3, and IQR, note: numeric_only=True
Q1 = df.quantile(0.25, numeric_only=True)
Q3 = df.quantile(0.75, numeric_only=True)
IQR = Q3 - Q1

print('Q1 =', Q1)
print('Q3 =', Q3)
print('IQR =', IQR)
# code ban đầu
# print("Q1 =", df.quantile (0.25, numeric_only=True))
# print("Q3 =", df.quantile (0.75,numeric_only=True))
# print("IQR=", (df.quantile (0.75,numeric_only=True))- (df.quantile (0.25, numeric_only=True)))
```

Don't worry about the above values because it's not important to know each and every one of them because it's just important to know how to use this technique in order to remove the outliers.

```python
# TODO: remove outliers using IQR (Xóa outlier bằng IQR),  note: numeric_only=True 
Q1 = df.quantile(0.25, numeric_only=True)
Q3 = df.quantile(0.75, numeric_only=True)
IQR = Q3 - Q1

df = df[(df[Q1.index] >= Q1 - 1.5 * IQR) & #df[Q1.index]: chỉ xử lý các cột số
        (df[Q1.index] <= Q3 + 1.5 * IQR)]

df.shape
```

```python
Q1 = df.quantile (0.25, numeric_only=True)
Q3 = df.quantile (0.75,numeric_only=True)
df = df.clip (lower = Q1, upper = Q3,axis = 1)
# clip(): ko xóa outlier, giữ lại dữ liệu nhưng giới hạn gtri
```

As seen above there were around 1600 rows were outliers. But you cannot completely remove the outliers because even after you use the above technique there maybe 1–2 outlier unremoved but that ok because there were more than 100 outliers. Something is better than nothing.

9.Plot different features against one another (scatter), against frequency (histogram)¶


**Histogram**

Histogram refers to the frequency of occurrence of variables in an interval. In this case, there are mainly 10 different types of car manufacturing companies, but it is often important to know who has the most number of cars. To do this histogram is one of the trivial solutions which lets us know the total number of car manufactured by a different company.

```python
# TODO: plot a bar chart of the top 40 car makes by number of cars nlargest(40)
# dịch: Vẽ biểu đồ cột thể hiện 40 thương hiệu xe hàng đầu dựa trên số lượng xe (sử dụng nlargest(40))
topcar = df['Make'].value_counts().nlargest(40) #nlargest(40): lấy 40 hãng nhiều xe nhất
plt.hist(topcar)
plt.show()
```

Heat Maps is a type of plot which is necessary when we need to find the dependent variables. One of the best way to find the relationship between the features can be done using heat maps.
```python
# TODO: compute correlation with df.corr(numeric_only=True), then plot a heatmap with sns.heatmap()
# dịch: Cần tính toán ma trận tương quan bằng df.corr(numeric_only=True), sau đó vẽ biểu đồ heatmap bằng sns.heatmap()
corr = df.corr(numeric_only=True)
sns.heatmap(corr, 
            annot = True, #hiện số tương quan trên từng ô
            cmap = 'coolwarm' #dùng màu để thể hiện mức độ tương quan
            ) #heatmap(): biểu diễn tương quan bằng màu
plt.show()
```

Create dummy variables for categorical columns with few levels. Scale numeric columns (keep the originals). Bin Price into Low / Medium / High.

```python
# TODO: use MinMaxScaler on Price and StandardScaler on HP (keep original columns (giữ lại các cột gốc))
# from sklearn.preprocessing import ?,?

# df['Price_minmax'] = ? (#tạo cột mới Price_minmax chứa giá Price sau khi chuẩn hóa bằng MinMaxScaler)
# df['HP_standard'] = ? (#tương tự trên)
# df[['Price', 'Price_minmax', 'HP', 'HP_standard']].head() (#hiển thị 3 cột để kt kq)
```

```python
# TODO: use pd.cut() to bin Price into Low / Medium / High (sd pd.cut() để chia Price thành 3 nhóm)
# df['Price_bin'] (#tạo cột mới tên Price_bin)

df['Price_bin'] = pd.cut(           #pd.cut(): chia dữ liệu số thành các khoảng
    df['MSRP'],                     #lấy giá xe để chia nhóm
    bins=[0, 30000, 60000,          #đặt các mốc chia
          float('inf')],            #float('inf'): ko giới hạn mức giá cao nhất
    labels=['Low', 'Medium', 'High']#đặt tên cho 3 nhóm
)
```
