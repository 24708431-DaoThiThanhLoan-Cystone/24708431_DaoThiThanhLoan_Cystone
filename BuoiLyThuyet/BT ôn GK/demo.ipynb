```python
# # Trả lời: (1đ)
# # Với mỗi cột thiếu, nên xóa dòng, xóa cột hay điền giá trị? Nêu lý do.
# MNAR 
# # Age: ? 
# MNAR (Missing Not At Random) - Dữ liệu thiếu không phải ngẫu nhiên, có thể liên quan đến giá trị của chính nó hoặc các biến khác. Ví dụ, nếu dữ liệu về tuổi bị thiếu vì một số người không muốn tiết lộ tuổi của họ, thì việc điền giá trị trung bình hoặc xóa dòng có thể làm mất thông tin quan trọng. Trong trường hợp này, nên xem xét các phương pháp khác như mô hình hóa dữ liệu thiếu hoặc sử dụng các kỹ thuật dự đoán để ước lượng giá trị thiếu.
# df = df['Age'].dropna() # xóa các dòng thiếu trong cột Age
df.dropna(subset={'Age'}, inplace=True) # xóa các dòng thiếu trong cột Age #tren bang
# # description: ?
# df = df['description'].fillna('Unknown', inplace=True) # điền giá trị 'Unknown' cho các dòng thiếu trong cột description

df.fillna(value = {'description': 'Unknown'}, inplace=True) # điền giá trị 'Unknown' cho các dòng thiếu trong cột description #tren bang

# # kiểm tra lại sau xử lý missing data
df.isnull().sum()
df[df['description'] == 'Unknown']
```

#### **Outlier (1.5đ)**

```python
# kiểm tra giá trị outlier 
# Tạo df3 bao gồm các cột Q1, Q3, IQR, lower_bound, upper_bound, outlier_count (0.5 đ)
# ?
df1 = df.select_dtypes(include='number')  # Chọn các cột số
Q1 = df1.quantile(0.25)
Q3 = df1.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_count = ((df1 < lower_bound) | (df1 > upper_bound)).sum() # ngưỡng vi phạm sau đó tính theo từng hàm 
# df3 = ?
df3 = pd.DataFrame({
    'Q1': Q1,
    'Q3': Q3,
    'IQR': IQR,
    'lower_bound': lower_bound,
    'upper_bound': upper_bound,
    'outlier_count': outlier_count
})
df3
```

```python
# Xử lý outlier bằng 2 cách: IQR và clip
# df4 cho IQR (0.5 đ)
df_clean = df1[~((df1 < lower_bound) | (df1 > upper_bound)).any(axis=1)]  # Loại bỏ các hàng có outlier
            #df chỗ này là df hay df1?
df_clean.shape
df_clean.head()
# df4 = df.copy()
# ?

# df5 cho clip (0.5 đ)
# df5 = df.copy()
# df5?
df5 = df.copy()
df5[df1.columns] = df5[df1.columns].clip(lower=lower_bound, 
                                         upper=upper_bound, 
                                         axis=1)
df5

print(df_clean.shape)
print(df5.shape)
# print(df4.shape)
# print(df5.shape)
```
