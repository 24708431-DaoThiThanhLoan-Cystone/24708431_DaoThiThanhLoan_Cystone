## Phần 3: Tạo cột mới và biến đổi dữ liệu


**Câu 10.** Từ cột `ORDERDATE`, tạo thêm 3 cột mới: `ORDER_YEAR`, `ORDER_MONTH`, `ORDER_QUARTER` chứa năm, tháng, quý của ngày đặt hàng.

```python
#đảm bảo ORDERDATE là kiểu ngày
df['ORDERDATE'] = pd.to_datetime(df['ORDERDATE'])
#df['ORDERDATE'] có thể thay bằng df.ORDERDATE
#pd.to_datetime(): dùng để chuyển dữ liệu sang kiểu ngày tháng mà Pandas có thể xử lý
```

```python
df.ORDERDATE.dt.year #lấy ngày tháng năm, đặc thù của ngày tháng năm phải có .dt
```

```python
#lấy năm
df['ORDER_YEAR'] = df['ORDERDATE'].dt.year # hoặc df['ORDER_YEAR'] = df.ORDERDATE.dt.year (cái này dễ nhìn hơn)
#.dt: cho phép lấy các thành phần của ngày tháng
#.year: lấy năm
```

```python
#lấy tháng
df['ORDER_MONTH'] = df['ORDERDATE'].dt.month
#.month: lấy tháng
```

```python
#lấy quý
df['ORDER_QUARTER'] = df['ORDERDATE'].dt.quarter
#.quarter: lấy quý
```

**Câu 11.** Tạo cột mới `PROFIT_MARGIN` = `Profit` / `SALES` (tỷ suất lợi nhuận). Đơn hàng nào có tỷ suất lợi nhuận cao nhất? Thấp nhất?

```python
#tạo cột tỷ suất lợi nhuận
df['PROFIT_MARGIN'] = df['Profit'] / df['SALES']
#nếu muốn hiển thị dạng % thì *100
```

```python
#tìm đơn có tỷ suất CAO nhất
df.loc[df['PROFIT_MARGIN'].idxmax()]
#idxmax(): tìm vị trí/index có giá trị lớn nhất
#loc[...]: lấy toàn bộ dòng ở vị trí đó
```

```python
#tìm đơn có tỷ suất THẤP nhất
df.loc[df['PROFIT_MARGIN'].idxmin()]
#idxmin(): tìm vị trí có giá trị nhỏ nhất
```

```python
#nếu chỉ muốn biết giá trị cao nhất/thấp nhất
df['PROFIT_MARGIN'].max()
df['PROFIT_MARGIN'].min()
```
