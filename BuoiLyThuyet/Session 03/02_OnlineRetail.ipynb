## A.4. Create a new column
Làm sạch dữ liệu (loại Quantity<=0, UnitPrice<=0), tạo cột `Sales` = Quantity * UnitPrice.

```python
# TODO
# lọc dữ liệu hợp lệ (loại Quantity<=0, UnitPrice<=0)
sale_clean = df[(df['Quantity']>0) & (df['UnitPrice'] >0)].copy() 
# .copy(): tạo một bản sao của dữ liệu đã lọc

# Sales = Quantity * UnitPrice
sale_clean['Sales'] = sale_clean['Quantity'] * sale_clean['UnitPrice']
sale_clean[['Quantity','UnitPrice','Sales']].head() #chỉ xem các cột nêu
```

# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

```python
# TODO
# Central Tendency = xu hướng trung tâm
# [mean(tbinh), media(trung vị), mode(gtri xuất hiện nhiều nhất)]
# Mean cho Sales
mean_sales = sale_clean['Sales'].mean()
print("**Mean:", mean_sales)
```

```python
# Median cho Sales
median_sales = sale_clean['Sales'].median()
print("**Median:", median_sales)
```

```python
# Mode cho Sales
mode_sales = sale_clean['Sales'].mode()[0]
print("**Mode:", mode_sales)
```

## Group 2 — Dispersion

```python
# TODO
#Dispersion (độ phân tán): để xem gtri Sales phân tán rộng hay tập trung quanh gtri tbinh
var_sales = sale_clean['Sales'].var() #.var(): tính phương sai 
std_sales = sale_clean['Sales'].std() #.std(): tính độ lệch chuẩn
range_sales = sale_clean['Sales'].max() - sale_clean['Sales'].min() #tính khoảng biến thiên

print("**Variance:", var_sales)
print("**Standard Deviation:", std_sales)
print("**Range:", range_sales)
```

## Group 3 — Location and Shape

```python
# TODO
# Location and Shape: xem vị trí của dữ liệu và hình dạng phân phối của dữ liệu
quantiles = sale_clean['Sales'].quantile([0.25, 0.5, 0.75]) #.quantile(): dùng để tìm các phân vị
skewness = sale_clean['Sales'].skew() #.skew(): đo xem phân phối dữ liệu lệch về bên nào
kurtosis = sale_clean['Sales'].kurtosis() #.kurtosis(): hình dạng đuôi và mức hiện gtri cực đoan

print("**Quantiles:\n", quantiles)
print("**Skewness:", skewness)
print("**Kurtosis:", kurtosis)
```

## Câu hỏi 1: Quốc gia nào đóng góp doanh thu cao nhất, chiếm bao nhiêu % tổng doanh thu?

```python
# TODO
country_sales=sale_clean.groupby('Country')['Sales'].sum().sort_values(ascending=False)
# chia dữ liệu theo từng country, cộng toàn bộ Sales của mõi country, sắp Sales từ lớn đến bé
print(country_sales.head(10),"\n")

# tính % Sales =  (country_sales / tổng country_sales) * 100
country_sales_percent = country_sales / country_sales.sum() * 100
print("**Country Sales Percent:\n", country_sales_percent)
```

## Câu hỏi 3: Doanh số có tính mùa vụ theo tháng không?

```python
# TODO
# tạo cột tháng
sale_clean['Month'] = sale_clean['InvoiceDate'].dt.month 
# .dt.month: lấy tháng từ cột ngày giờ

# tính doanh thu theo tháng
monthly_sales = sale_clean.groupby('Month')['Sales'].sum()

# tạo biểu đồ
plt.figure(figsize=(10,5)) #tạo khung (rộng 10, cao 5)
sns.lineplot(x=monthly_sales.index, y=monthly_sales.values) #vẽ đường
plt.title("Monthly Sales Trend") #tiêu đề
plt.xlabel("Month") #tên trục X
plt.ylabel("Sales") #tên trục Y
plt.show() #hiện lên
```

## Câu hỏi 4: Giá trị đơn hàng trung bình (Average Order Value) khác nhau thế nào giữa các quốc gia?

```python
# TODO
order_value = sale_clean.groupby('Country').apply(lambda g: g['Sales'].sum() / g['InvoiceNo'].nunique())
# chia dữ liệu theo từng nhóm Country 
# .apply(lambda g:...): phép tính riêng cho từng country 
# g['Sales'].sum(): tính tổng doanh thu của quốc gia đó
# g['InvoiceNo'].nunique(): đếm số mã hóa đơn khác nhau
order_value.sort_values(ascending=False).head(10) #sắp từ lớn đến bé
```

## Câu hỏi 5: Tỷ lệ giao dịch có dấu hiệu trả hàng/hủy (Quantity âm ở dữ liệu gốc) khác nhau thế nào giữa các quốc gia?

```python
# TODO
# đếm số đơn trả hàng theo Country
returns = df[df['Quantity'] < 0].groupby('Country')['InvoiceNo'].nunique()

# đếm tổng số đơn hàng
total_orders = df.groupby('Country')['InvoiceNo'].nunique()

# tính tỷ lệ trả hàng
return_rate = (returns / total_orders * 100).fillna(0).sort_values(ascending=False)
# tỷ lệ trả hàng = (số đơn có trả hàng / tổng số đơn) × 100
# .fillna(0): gặp ô nào bị NaN (thiếu dữ liệu) → thay bằng 0
return_rate.head(10)
```
